```python
# An implementation of neural_network (MLPClassifier)

import numpy as np
import pandas as pd
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, confusion_matrix
# Location of dataset
from google.colab import drive

# Mount Google Drive
drive.mount('/content/gdrive')

# Load the dataset
dataset = pd.read_csv('gdrive/My Drive/ai lab/earthquake_1995-2023(1000).csv')



dataset.head()
#select the columns from the dataset and assign them to the variable
X = dataset[['magnitude',	'cdi',	'mmi', 'sig', 'nst',	'dmin',	'gap',  'depth',	'latitude',	'longitude']]
#the target
y = dataset['tsunami']


# Assign data from first fifth columns to y variable
y.head()

#spliting 0.20, which means that 20% of the data will be used for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20)

#standardize the input features
scaler = StandardScaler()
scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

#specifies the number of neurons in each hidden layer of the neural network, specifies the maximum number of iterations for the solver to converge
mlp = MLPClassifier(hidden_layer_sizes=(10, 10, 10), max_iter=1000)
mlp.fit(X_train, y_train.values.ravel())

#make predictions
predictions = mlp.predict(X_test)


print(confusion_matrix(y_test,predictions))
print(classification_report(y_test,predictions))
```
